## 1. 提示词模板

举例1： 提示词模板 PromptTemplate.from_template

```python
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

from openai import base_url

load_dotenv(override=True)

#初始化模型
model = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=os.getenv('DASHSCOPE_API_KEY'),
    base_url=os.getenv('DASHSCOPE_BASE_URL'),
)

topic = "Python"
difficulty = "初学者"

template = PromptTemplate.from_template(
    f"你是一个{difficulty}级别的编程导师。请用简单易懂的语言解释{topic}"
)

# 使用模板生成提示词
prompt = template.format(difficulty=difficulty, topic=topic)
#调用提示词模板，并且返回结果
response = model.invoke(prompt)

print(response.content)
```

## 2. 模板ChatPromptTemplate的三种调用方法：invoke() 、 format() 、 format_messages()

举例1；invoke() 返回ChatPromptValue


```python
from langchain_core.prompts import ChatPromptTemplate

# 元组构成的列表
prompt_template = ChatPromptTemplate(
    [
    # 字符串 role + 字符串 content
    ("system", "你是一个AI开发工程师. 你的名字是 {name}."),
    ("human", "你能开发哪些AI应用?"),
    ("ai", "我能开发很多AI应用, 比如聊天机器人, 图像识别, 自然语言处理等."),
    ("human", "{user_input}")
    ]
)
#返回ChatPromptValue
prompt = prompt_template.invoke({'name': '小顾AI', 'user_input': '你能帮我做什么？'})

print(type(prompt))

print(prompt)

print(len(prompt.messages))
```

## 3. 结合LLM调用

举例：

```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os

load_dotenv(override=True)

model = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=os.getenv('DASHSCOPE_API_KEY'),
    base_url=os.getenv('DASHSCOPE_BASE_URL'),
)

chat_template = ChatPromptTemplate.from_messages(
    [
        ('system', '你是一个数学家，你可以计算任何算式'),
        ('human', '{msg_text}'),
    ]
)

#提示词模板的变量值赋值
prompt = chat_template.format_messages(msg_text='我今年18岁，我的舅舅今年38岁，我的爷爷今年72岁，我和舅舅一共多少岁了？')

#调用提示词
response = model.invoke(prompt)

print(response.content)
```

## 4. 提示词模板的初始化参数类型
类型1：str列表类型

列表参数格式是str类型（不推荐），因为默认角色都是human

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [
        "Hello, {name}"
    ]
)

message = chat_template.invoke({"name": "小智AI"})

print(type(message))

print(message)
```

类型2：tuple列表类型

列表参数格式是元组类型


```python
from langchain_core.prompts import ChatPromptTemplate

#元组形式的消息
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "你的名字是{role}"),
        ("human", "很高兴认识你")
    ]
)

print(prompt.invoke({"role": "小智"}))
```

类型3：dict（字典）列表类型

列表参数格式是dict类型

```python
from langchain_core.prompts import ChatPromptTemplate

#字典形式的消息
prompt = ChatPromptTemplate.from_messages(
    [
        {"role": "system", "content": "你的名字是{role}"},
        {"role": "human", "content": "很高兴认识你"}
    ]
)

print(prompt.invoke({"role": "小智"}))
```

类型4：Message列表类型

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage, HumanMessage

chat_template = ChatPromptTemplate.from_messages(
    [
        SystemMessage(content="我是一个贴心的只能助手"),
        HumanMessage(content="我的问题是：{word}英文怎么说？")
    ]
)

messages = chat_template.invoke({"word":"人工智能"})

print(messages)

print(type(messages))
```

类型5：MessagePromptTemplate列表类型

LangChain提供不同类型的MessagePromptTemplate。最常用的是SystemMessagePromptTemplate 、 HumanMessagePromptTemplate 和 AIMessagePromptTemplate ，分别创建系统消息、人工消息和AI消息。

```python
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate

system_msg_prompt = SystemMessagePromptTemplate.from_template("你是一个{role}")
human_msg_prompt = HumanMessagePromptTemplate.from_template("给我解释{concept},用浅显易懂的语言")

chat_prompt = ChatPromptTemplate.from_messages(
    [
        system_msg_prompt,
        human_msg_prompt
    ]
)

#格式化提示
format_messages = chat_prompt.invoke({"role": "物理学家", "concept": "相对论"})

print(format_messages)
```

类型6：BaseChatPromptTemplate列表类型

使用 BaseChatPromptTemplate，可以理解为ChatPromptTemplate里嵌套了ChatPromptTemplate。

```python
from langchain_core.prompts import (
    ChatPromptTemplate,SystemMessagePromptTemplate,HumanMessagePromptTemplate,
)
from langchain.messages import SystemMessage, HumanMessage
#示例1： 使用BaseMessage（已实例化的消息）
system_msg = SystemMessage(content="你是一个AI工程师。")
human_msg = HumanMessage(content="你好！")

#示例2：使用BaseMessagePromptTemplate
system_prompt = SystemMessagePromptTemplate.from_template("你是一个{role}。")
human_prompt = HumanMessagePromptTemplate.from_template("{user_input}")

#示例3：使用BaseChatPromptTemplate（嵌套的ChatPromptTemplate）
nested_prompt = ChatPromptTemplate.from_messages([("system", "嵌套提示词")])

prompt_template = ChatPromptTemplate.from_messages(
   [
    system_msg,     # MessageLike （BaseMessage）
    human_msg,      # MessageLike （BaseMessage）
    system_prompt,  # MessageLike （BaseMessagePromptTemplate）
    human_prompt,   # MessageLike （BaseMessagePromptTemplate）
    nested_prompt   # MessageLike （BaseChatPromptTemplate）
   ]
)

prompt_template.invoke({"role": "人工智能专家", "user_input": "介绍以下大模型的应用场景"})
```

## 5. 高级特性
### 5.1  部分变量预填充：partial()

预填充某些固定不变的变量，创建模板的变体。

使用场景：
- 某些变量在所有调用中都相同
- 需要为不同用户/场景创建定制模板

举例：

```python
from langchain_core.prompts import ChatPromptTemplate

# 原始模板，为不同部门的创建专业模板
base_template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是{department}的{role}"),
        ("user", "{task}")
    ]
)

# IT 部门
it_template = base_template.partial(
    department="IT 部门",
    role="技术支持"
)

#销售部门
sales_template = base_template.partial(
    department="销售部门",
    role="销售顾问"
)

sales_template.invoke({"task":"为什么每年年底汽车会促销？"})
```

### 5.2 消息占位符
当你不确定消息提示模板使用什么角色，或者希望在格式化过程中 插入消息列表 时，该怎么办？ 这就需要使用消息占位符，负责在特定位置添加消息列表。

使用场景：多轮对话系统存储历史消息以及Agent的中间步骤处理此功能非常有用。

举例1：JSON格式的占位符

```python
from langchain_core.prompts import ChatPromptTemplate

template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个有用的AI助手"),
        ("placeholder", "{conversion}")
    ]
)

prompt_value = template.invoke(
    {
        "conversion": [
            ("human", "你好！"),
            ("ai", "今天我能帮你做什么？"),
            ("human", "你能给我一个冰淇淋吗？"),
            ("ai", "抱歉，我没有这样子的能力")
        ]
    }
)

print(prompt_value)
```

MessagesPlaceholder实例

举例2：简单的

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage

prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant"),
        MessagesPlaceholder("msgs")
    ]
)

prompt_template.invoke({"msgs": [HumanMessage(content="hi")]})

```

举例3：

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)

model = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=os.getenv('DASHSCOPE_API_KEY'),
    base_url=os.getenv('DASHSCOPE_BASE_URL')
)

prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个非常友好的AI助手"),
        MessagesPlaceholder(variable_name="history"),
        ("human", "{question}")
    ]
)

prompt = prompt_template.invoke(
    {
        "history": [
            ("human", "5 + 2 = ?"),
            ("ai", "5 + 2 = 7")
        ],
        "question": "结果再乘以4呢？"
    }
)

result = model.stream(prompt)

for chunk in result:
    print(chunk.text, end='', flush=True)
```
