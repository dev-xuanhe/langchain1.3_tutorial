## 1. 提示词模板

举例1： 提示词模板 PromptTemplate.from_template

```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

load_dotenv(override=True)

#初始化模型
model = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=os.getenv('DASHSCOPE_API_KEY'),
    base_url=os.getenv('DASHSCOPE_BASE_URL'),
)

topic = "Python"
difficulty = "初学者"

template = PromptTemplate.from_template(
    f"你是一个{difficulty}级别的编程导师。请用简单易懂的语言解释{topic}"
)

# 使用模板生成提示词
prompt = template.format(difficulty=difficulty, topic=topic)
#调用提示词模板，并且返回结果
response = model.invoke(prompt)

print(response.content)
```

## 2. 模板ChatPromptTemplate的三种调用方法：invoke() 、 format() 、 format_messages()

举例1；invoke() 返回ChatPromptValue


```python
from langchain_core.prompts import ChatPromptTemplate

# 元组构成的列表
prompt_template = ChatPromptTemplate(
    [
    # 字符串 role + 字符串 content
    ("system", "你是一个AI开发工程师. 你的名字是 {name}."),
    ("human", "你能开发哪些AI应用?"),
    ("ai", "我能开发很多AI应用, 比如聊天机器人, 图像识别, 自然语言处理等."),
    ("human", "{user_input}")
    ]
)
#返回ChatPromptValue
prompt = prompt_template.invoke({'name': '小顾AI', 'user_input': '你能帮我做什么？'})

print(type(prompt))

print(prompt)

print(len(prompt.messages))
```

举例2：format() 返回string


```python
from langchain_core.prompts import ChatPromptTemplate

# 元组构成的列表
prompt_template = ChatPromptTemplate(
    [
    # 字符串 role + 字符串 content
    ("system", "你是一个AI开发工程师. 你的名字是 {name}."),
    ("human", "你能开发哪些AI应用?"),
    ("ai", "我能开发很多AI应用, 比如聊天机器人, 图像识别, 自然语言处理等."),
    ("human", "{user_input}")
    ]
)

prompt = prompt_template.format(name='小顾AI', user_input='你能帮我做什么？')

print(type(prompt))

print(prompt)
```

举例3：format_messages()返回消息列表


```python
from langchain_core.prompts import ChatPromptTemplate

# 元组构成的列表
prompt_template = ChatPromptTemplate(
    [
    # 字符串 role + 字符串 content
    ("system", "你是一个AI开发工程师. 你的名字是 {name}."),
    ("human", "你能开发哪些AI应用?"),
    ("ai", "我能开发很多AI应用, 比如聊天机器人, 图像识别, 自然语言处理等."),
    ("human", "{user_input}")
    ]
)

prompt = prompt_template.format_messages(name='小顾AI', user_input='你能帮我做什么？')

print(type(prompt))

print(prompt)
```

## 3. 结合LLM调用

举例：

```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from dotenv import load_dotenv
import os

load_dotenv(override=True)

model = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=os.getenv('DASHSCOPE_API_KEY'),
    base_url=os.getenv('DASHSCOPE_BASE_URL'),
)

chat_template = ChatPromptTemplate.from_messages(
    [
        ('system', '你是一个数学家，你可以计算任何算式'),
        ('human', '{msg_text}'),
    ]
)

#提示词模板的变量值赋值
prompt = chat_template.format_messages(msg_text='我今年18岁，我的舅舅今年38岁，我的爷爷今年72岁，我和舅舅一共多少岁了？')

#调用提示词
response = model.invoke(prompt)

print(response.content)
```
