# 流式调用
invoke 和 stream 有什么区别？
- invoke() ：同步调用，在模型输出完成后一次性获取响应，对于输出文本很长的场景，用户体验不好。
- stream() ：流式调用，实时返回响应片段。调用后，返回一个 迭代器(iterator) ，可以通过循环来实时处理每一个新生成的chunk内容块。

注意：流式输出依赖于模型供应商对于流式输出的支持

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
import os

load_dotenv(override=True)

API_KEY = os.getenv("DASHSCOPE_API_KEY")
BASE_URL = os.getenv("DASHSCOPE_BASE_URL")

model_openai = init_chat_model(
    model='qwen3.7-flash',
    model_provider='openai',
    api_key=API_KEY,
    base_url=BASE_URL,
)

for chunk in model_openai.stream('写一首七言律诗，总结大模型的发展'):
    print(chunk.text, end='', flush=True)  #逐个token输出
```
